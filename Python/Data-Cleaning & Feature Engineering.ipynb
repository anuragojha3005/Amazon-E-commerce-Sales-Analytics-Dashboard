# Feature Engineering

```python
df['discount_amount'] = df["price"] - df["final_price"]
```

```python
df[["price", "discount", "final_price", "discount_amount"]].head()
```

```python
df["saving_amount"] = (df["discount_amount"] / df["price"])*100
df["saving_amount"]
```

```python
def price_category(price):
    if price < 1000:
        return "Budget"
    elif price < 5000:
        return "Mid Range"
    else:
        return "Premium"

df["price_category"] = df["price"].apply(price_category)
```

```python
df["price_category"].value_counts()
```

```python
def rating_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4:
        return "Good"
    elif rating >= 3:
        return "Average"
    else:
        return "Poor"

df["rating_category"] = df["rating"].apply(rating_category)
```

```python
df["rating_category"].value_counts()
```

```python
def shipping_speed(days):
    if days <= 2:
        return "Fast"
    elif days <= 5:
        return "Standard"
    else:
        return "Slow"

df["shipping_speed"] = df["shipping_time_days"].apply(shipping_speed)

df["shipping_speed"].value_counts()
```

```python
def stock_status(stock):
    if stock < 20:
        return "Low"
    elif stock < 100:
        return "Medium"
    else:
        return "High"

df["stock_status"] = df["stock"].apply(stock_status)

df["stock_status"].value_counts()
```

```python
df.columns
```

```python
def review_category(review):
    if review < 50:
        return "Low"
    elif review < 500:
        return "Medium" 
    else:
        return "High"

df["review_category"] = df["review_count"].apply(review_category)

```

```python
df["is_weekend"] = df["Day"].isin(["Saturday", "Sunday"])
```

```python
df["high_value_order"] = df["final_price"] > df["final_price"].mean()
```

```python
def return_risk(days, rating):
    if days > 5 and rating < 3.5:
        return "High"
    elif days > 3:
        return "Medium"
    else:
        return "Low"

df["return_risk"] = df.apply(
    lambda x: return_risk(
        x["shipping_time_days"],
        x["rating"]
    ),
    axis=1
)
```

```python
def seller_performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4:
        return "Good"
    else:
        return "Average"

df["seller_performance"] = df["seller_rating"].apply(seller_performance)
```

```python
df["delivery_status"].value_counts()
```

```python
def delivery_quality(status):
    if status == "Delivered":
        return "Successful"
    else:
        return "Issue"

df["delivery_quality"] = df["delivery_status"].apply(delivery_quality)
```
